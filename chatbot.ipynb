```python
import google.generativeai as genai
from dotenv import load_dotenv
import os

load_dotenv()

API_KEY = os.getenv("GEMINI_API_KEY")  # 👈 match .env name

if not API_KEY:
    raise ValueError("GEMINI_API_KEY not found in .env file")

genai.configure(api_key=API_KEY)

model = genai.GenerativeModel("gemini-2.5-flash")
response = model.generate_content("Explain AI in one sentence")
print(response.text)

subject_input = widgets.Text(
    description="Subject:",
    layout=widgets.Layout(width='400px')
)
question_input = widgets.Text(
    description="Question:",
    layout=widgets.Layout(width='400px')
)
description_input = widgets.Dropdown(
    description="Type:",
    options=['MCQ', 'one sentence', 'Explain', 'descriptive'],
    layout=widgets.Layout(width='400px')
)
generate_button = widgets.Button(
    description="Generate Answer",
    button_style='success',
    layout=widgets.Layout(width='400px')
)

output = widgets.Output()

def generate_answer(b):
    output.clear_output()

    subject = subject_input.value.strip()
    question = question_input.value.strip()
    question_type = description_input.value

    if not subject or not question:
        with output:
            print("Please enter both subject and question.")
        return


    if question_type == "MCQ":
        instruction = "Generate at least 10 multiple choice questions (with 4 options each) on the topic."
    elif question_type == "one sentence":
        instruction = "Answer the question in one precise sentence."
    elif question_type == "Explain":
        instruction = "Explain the concept in detail with examples if relevant."
    else:
        instruction = "Write a descriptive answer covering all necessary aspects."

    # Build prompt
    prompt = f"""
You are an AI Education Assistant. Use the following parameters:

- Subject: {subject}
- Question: {question}
- Question Type: {question_type}

Instruction:
{instruction}

Ensure clarity, appropriate academic tone, and proper formatting.
"""

    with output:
        try:
            response = model.generate_content(prompt)
            display(Markdown(f"### 📘 Generated Answer:\n\n{response.text}"))
        except Exception as e:
            print(f"An error occurred: {e}")

generate_button.on_click(generate_answer)

form = widgets.VBox([
    subject_input,
    question_input,
    description_input,
    generate_button,
    output
])
display(form)
```
